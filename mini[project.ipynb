```python
with open('tim.txt','r') as file :
  codes=file.readlines()
```

```python
print(codes)
```

```python
for line in codes:
  #strip to clean those extra character i.e \n check with the above code 87 or above output
  clean_line=line.strip()


  if "ERROR" in clean_line:
   parts= clean_line.split()
   time_stamp=f'{parts[0]},{parts[1]}'
   print(time_stamp)

  Error_logs.append(time_stamp)
```
